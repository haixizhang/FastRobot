```python
%load_ext autoreload
%autoreload 2

from ble import get_ble_controller
from base_ble import LOG
from cmd_types import CMD
import time
import numpy as np
import matplotlib.pyplot as plt
LOG.propagate = False
```

```python
# Get ArtemisBLEController object
ble = get_ble_controller()
# Connect to the Artemis Device
ble.connect()
```

```python
# Global lists to store the PID+ToF+Motor data received from the Arduino.
pid_time = []      # Timestamps (ms)
pid_error = []     # PID error (setpoint - distance)
pid_output = []    # PID controller output (before mapping to motor command)
pid_distance = []  # Raw ToF sensor distance reading (mm)
pid_motor = []     # Motor PWM command sent to the motors
notification_count = 0

def pid_notification_handler(sender: int, data: bytearray):
    #Expected message format for PID data: "T:<time>|E:<error>|U:<output>|D:<distance>|M:<motorPWM>"
    global notification_count
    try:
        message = data.decode("utf-8").strip()
        # Only parse if the message starts with the PID data prefix
        if message.startswith("T:"):
            parts = message.split("|")
            if len(parts) < 5:
                print(f"Incomplete PID data: {message}")
                return
            t = float(parts[0].split(":")[1])
            e = float(parts[1].split(":")[1])
            u = float(parts[2].split(":")[1])
            d = float(parts[3].split(":")[1])
            m = float(parts[4].split(":")[1])
            pid_time.append(t)
            pid_error.append(e)
            pid_output.append(u)
            pid_distance.append(d)
            pid_motor.append(m)
            notification_count += 1
            print(f"PID Data Notification #{notification_count}: {message}")
        else:
            print(f"Ack or non-PID notification received: {message}")
    except Exception as ex:
        print("Error parsing notification:", ex)
```

```python
ble.start_notify(ble.uuid['RX_STRING'], pid_notification_handler)
```

```python
# Step 1: Set PID Gains
ble.send_command(CMD.SET_PID_GAINS, "0.3|0.001|0|304")
time.sleep(1)  # Pause briefly to allow the device to update its parameters
```

```python
# Step 2: Start PID Control Test
# Instruct the device to run PID control for 5000 ms (5 seconds).
ble.send_command(CMD.START_PID_CONTROL, "5000")
# Wait a bit longer than the test duration to ensure the test completes.
```

```python
# Step 3: Request Logged PID+ToF Data
ble.send_command(CMD.GET_PID_DATA, "")
# Allow time for the device to send back all notifications.
time.sleep(5)
```

```python
# Now plot the received data in three subplots
plt.figure(figsize=(12, 12))

# Plot PID Error vs. Time
plt.subplot(3, 1, 1)
plt.plot(pid_time, pid_error, 'bo-', label='PID Error')
plt.xlabel('Time (ms)')
plt.ylabel('Error (mm)')
plt.title('PID Error vs. Time')
plt.legend()
plt.grid(True)

# Plot ToF Distance vs. Time
plt.subplot(3, 1, 2)
plt.plot(pid_time, pid_distance, 'ro-', label='ToF Distance')
plt.xlabel('Time (ms)')
plt.ylabel('Distance (mm)')
plt.title('ToF Distance vs. Time')
plt.legend()
plt.grid(True)

# Plot Motor PWM vs. Time
plt.subplot(3, 1, 3)
plt.plot(pid_time, pid_motor, 'ko-', label='Motor PWM')
plt.xlabel('Time (ms)')
plt.ylabel('PWM Value')
plt.title('Motor PWM vs. Time')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Get ArtemisBLEController object
ble = get_ble_controller()
# Connect to the Artemis Device
ble.connect()
```

```python
# Global lists to store logged PID+ToF+Motor data received from the Arduino.
pid_time = []              # Timestamps (ms)
pid_error = []             # PID error (setpoint - estimated distance)
pid_output = []            # PID controller output (before mapping)
pid_raw_distance = []      # Raw ToF sensor distance (mm)
pid_estimated_distance = []# Extrapolated (estimated) distance (mm)
pid_motor_pwm = []         # Motor PWM command that was sent

notification_count = 0

def notification_handler(sender: int, data: bytearray):
    """
    Callback to handle notifications from the Arduino.
    Expected message format: 
    "T:<time>|E:<error>|U:<output>|R:<rawDist>|X:<estimatedDist>|M:<motorPWM>"
    """
    global notification_count
    try:
        message = data.decode("utf-8").strip()
        # Check if the message starts with the expected PID data token
        if message.startswith("T:"):
            parts = message.split("|")
            t = float(parts[0].split(":")[1])
            e = float(parts[1].split(":")[1])
            u = float(parts[2].split(":")[1])
            r = float(parts[3].split(":")[1])
            x = float(parts[4].split(":")[1])
            m = float(parts[5].split(":")[1])
            
            pid_time.append(t)
            pid_error.append(e)
            pid_output.append(u)
            pid_raw_distance.append(r)
            pid_estimated_distance.append(x)
            pid_motor_pwm.append(m)
            
            notification_count += 1
            print(f"Notification #{notification_count}: {message}")
        else:
            print(f"Received non-PID data: {message}")
    except Exception as ex:
        print("Error parsing notification:", ex)

ble.start_notify(ble.uuid['RX_STRING'], notification_handler)
```

```python
pid_time = []
pid_error = []
pid_output = []
pid_raw_dist = []
pid_extrap_dist = []
pid_motor = []
notification_count = 0

def pid_notification_handler(sender: int, data: bytearray):
    """
    Callback to handle notifications sent from the Arduino.
    Expected message format:
      "T:<time>|E:<error>|U:<output>|R:<raw_dist>|X:<extrap_dist>|M:<motorPWM>"
    """
    global notification_count
    try:
        message = data.decode("utf-8").strip()
        
        # Check if the message starts with the PID data prefix "T:"
        if message.startswith("T:"):
            parts = message.split("|")
            # We expect 6 parts total: T, E, U, R, X, M
            if len(parts) < 6:
                print(f"Incomplete PID data: {message}")
                return
            
            t_str = parts[0].split(":")[1]  # after 'T:'
            e_str = parts[1].split(":")[1]  # after 'E:'
            u_str = parts[2].split(":")[1]  # after 'U:'
            r_str = parts[3].split(":")[1]  # after 'R:'
            x_str = parts[4].split(":")[1]  # after 'X:'
            m_str = parts[5].split(":")[1]  # after 'M:'
            
            # Convert to float (or int where appropriate)
            t_val = float(t_str)
            e_val = float(e_str)
            u_val = float(u_str)
            r_val = float(r_str)
            x_val = float(x_str)
            m_val = float(m_str)
            
            # Store in global lists
            pid_time.append(t_val)
            pid_error.append(e_val)
            pid_output.append(u_val)
            pid_raw_dist.append(r_val)
            pid_extrap_dist.append(x_val)
            pid_motor.append(m_val)
            
            notification_count += 1
            print(f"PID Data Notification #{notification_count}: {message}")
        else:
            # If it doesn't start with 'T:', treat as an ACK or debugging message
            print(f"Ack or other notification: {message}")
    except Exception as ex:
        print("Error parsing notification:", ex)


ble.start_notify(ble.uuid['RX_STRING'], pid_notification_handler)
```

```python
t_sec = (np.array(pid_time) - pid_time[0]) / 1000.0  # zero-based, in seconds

fig, axs = plt.subplots(4, 1, figsize=(10, 14))

# Subplot 1: PID Error vs. Time
axs[0].plot(t_sec, pid_error, 'b-', label='PID Error')
axs[0].set_xlabel("Time (s)")
axs[0].set_ylabel("Error (mm)")
axs[0].set_title("PID Error vs. Time")
axs[0].grid(True)
axs[0].legend()

# Subplot 2: Raw & Extrapolated Distance vs. Time
axs[1].plot(t_sec, pid_raw_dist, 'r-', label='Raw Distance')
axs[1].plot(t_sec, pid_extrap_dist, 'g-', label='Extrapolated Distance')
axs[1].set_xlabel("Time (s)")
axs[1].set_ylabel("Distance (mm)")
axs[1].set_title("Raw vs. Extrapolated Distance")
axs[1].grid(True)
axs[1].legend()

# Subplot 3: PID Output (before PWM mapping) vs. Time
axs[2].plot(t_sec, pid_output, 'm-', label='PID Output')
axs[2].set_xlabel("Time (s)")
axs[2].set_ylabel("Controller Output")
axs[2].set_title("PID Output vs. Time")
axs[2].grid(True)
axs[2].legend()

# Subplot 4: Motor PWM vs. Time
axs[3].plot(t_sec, pid_motor, 'k-', label='Motor PWM')
axs[3].set_xlabel("Time (s)")
axs[3].set_ylabel("PWM Value")
axs[3].set_title("Motor PWM vs. Time")
axs[3].grid(True)
axs[3].legend()

plt.tight_layout()
plt.show()
```

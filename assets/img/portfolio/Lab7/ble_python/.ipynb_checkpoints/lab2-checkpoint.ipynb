```python
%load_ext autoreload
%autoreload 2

from ble import get_ble_controller
from base_ble import LOG
from cmd_types import CMD
import time
import numpy as np
import matplotlib.pyplot as plt
LOG.propagate = False
```

```python
LOG.debug("debug")
LOG.info("info")
LOG.warning("warning")
LOG.error("error")
LOG.critical("critical")
```

```python
# Get ArtemisBLEController object
ble = get_ble_controller()

# Connect to the Artemis Device
ble.connect()
```

```python
# 1
# Global lists to store time stamps (ms) and data (pitch and roll values)
data_times = []
pitch_data = []
roll_data = []
notification_count = 0

def notification_handler(sender: int, data: bytearray):
    global notification_count
    try:
        # Decode the received bytes to a UTF-8 string
        message = data.decode("utf-8")
        # Expected message format: "P:<pitch>, R:<roll>"
        parts = message.split(',')
        print(message)
        pitch_str = parts[0].split(':')[1].strip()  # extract pitch value
        roll_str = parts[1].split(':')[1].strip()   # extract roll value
        # Convert to float
        pitch_val = float(pitch_str)
        roll_val = float(roll_str)
        current_time = time.time()
        # Append values to global lists
        data_times.append(current_time)
        pitch_data.append(pitch_val)
        roll_data.append(roll_val)
        notification_count += 1
        print(f"Notification {notification_count} - Time: {current_time:.0f} ms, Pitch: {pitch_val:.2f}°, Roll: {roll_val:.2f}°")
    except Exception as e:
        print(f"Error parsing notification: {e}")
```

```python
# Start notifications on the characteristic for pitch/roll (adjust the UUID as needed)
ble.start_notify(ble.uuid['RX_STRING'], notification_handler)
for i in range(150):
    # Send a command to request pitch/roll data.
    ble.send_command(CMD.GET_PITCH_ROLL, "")
```

```python
ble.stop_notify(ble.uuid['RX_STRING'])
print(f"\nTotal notifications received: {len(data_times)}")
```

```python
# Plot the data using matplotlib.
def plot_pitch_roll(time_data, pitch_values, roll_values):
    plt.figure(figsize=(10, 6))
    # Plot pitch vs time
    plt.subplot(2, 1, 1)
    plt.plot(time_data, pitch_values, label='Pitch (deg)', marker='o', linestyle='-')
    plt.xlabel('Time (ms)')
    plt.ylabel('Pitch (°)')
    plt.title('Pitch vs. Time')
    plt.grid(True)
    plt.legend()
    # Plot roll vs time
    plt.subplot(2, 1, 2)
    plt.plot(time_data, roll_values, label='Roll (deg)', color='orange', marker='o', linestyle='-')
    plt.xlabel('Time (ms)')
    plt.ylabel('Roll (°)')
    plt.title('Roll vs. Time')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
# Call the plotting function.
plot_pitch_roll(data_times, pitch_data, roll_data)
```

```python
def plot_fft(data, time_data, title="Frequency Spectrum", xlabel="Frequency (Hz)", ylabel="Amplitude"):
    # Convert lists to numpy arrays
    time_array = np.array(time_data)
    data_array = np.array(data)
    time_sec = time_array 
    # Calculate the sampling interval and frequency
    dt = np.mean(np.diff(time_sec))
    fs = 1.0 / dt  # Sampling frequency in Hz
    print(f"Sampling frequency: {fs:.2f} Hz")
    # Number of sample points
    n = len(data_array)
    # Compute FFT
    fft_data = np.fft.fft(data_array)
    # Only take the positive frequency components
    fft_data = fft_data[:n//2]
    # Compute corresponding frequency bins
    freqs = np.fft.fftfreq(n, d=dt)[:n//2]
    # Normalize and compute magnitude
    magnitude = np.abs(fft_data) / n
    # Plot the frequency spectrum
    plt.figure(figsize=(10, 6))
    plt.plot(freqs, magnitude, marker='o', color = 'orange')
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True)
    plt.show()
```

```python
plot_fft(pitch_data, data_times, title="Pitch Data Frequency Spectrum")
```

```python
plot_fft(roll_data, data_times, title="Roll Data Frequency Spectrum")
```

```python
def lowpass_filter(data, cutoff, fs):
    #Apply a simple single-pole IIR lowpass filter to the data.
    data = np.array(data)
    dt = 1.0 / fs
    RC = 1.0 / (2 * np.pi * cutoff)
    alpha = dt / (RC + dt)
    filtered = np.zeros_like(data)
    filtered[0] = data[0]
    for i in range(1, len(data)):
        filtered[i] = filtered[i-1] + alpha * (data[i] - filtered[i-1])
    return filtered

# Calculate the sampling frequency from our data_times (converted to seconds)
time_array = np.array(data_times)
dt = np.mean(np.diff(time_array))
fs = 1.0 / dt
print(f"Calculated Sampling Frequency: {fs:.2f} Hz")

# Choose a cutoff frequency based on the FFT analysis.
cutoff_freq = 2.0  # in Hz

# Apply the lowpass filter to the pitch data
filtered_pitch = lowpass_filter(pitch_data, cutoff_freq, fs)

# Plot the original and filtered signals versus time
plt.figure(figsize=(12, 7))
plt.plot(data_times, pitch_data, label='Original Signal', color='blue', alpha=0.5)
plt.plot(data_times, filtered_pitch, label='Filtered Signal (Cutoff = 50 Hz)', color='red', linewidth=2)
plt.xlabel("Time (ms)")
plt.ylabel("Pitch (°)")
plt.title("Original vs. Filtered Pitch Signal")
plt.legend()
plt.grid(True)
plt.show()
```

```python
time_data = []
gyro_pitch_data = []
gyro_roll_data  = []
gyro_yaw_data   = []
acc_pitch_data  = []
acc_roll_data   = []
start_time = None
def gyro_notification_handler(sender, data):
    global start_time
    message = data.decode('utf-8').strip()
    try:
        tokens = message.split(';')
        gyro_pitch = float(tokens[0].split(':')[1].strip())
        gyro_roll  = float(tokens[1].split(':')[1].strip())
        gyro_yaw   = float(tokens[2].split(':')[1].strip())
        acc_pitch  = float(tokens[3].split(':')[1].strip())
        acc_roll   = float(tokens[4].split(':')[1].strip())
        if start_time is None:
            start_time = time.time()
        t = time.time() - start_time
        time_data.append(t)
        gyro_pitch_data.append(gyro_pitch)
        gyro_roll_data.append(gyro_roll)
        gyro_yaw_data.append(gyro_yaw)
        acc_pitch_data.append(acc_pitch)
        acc_roll_data.append(acc_roll)
    except Exception as e:
        print("Error parsing gyro notification:", e)
        print("Raw message was:", message)
```

```python
ble.start_notify(ble.uuid['RX_STRING'], gyro_notification_handler)
for _ in range(150):
    ble.send_command(CMD.GET_FROM_GYROSCOPE, "") 
```

```python
# ble.stop_notify(ble.uuid['RX_STRING'])
plt.figure(figsize=(10,8))
plt.subplot(3,1,1)
plt.plot(time_data, gyro_pitch_data, 'b', label='Gyro Pitch')
plt.plot(time_data, acc_pitch_data,  'b--', label='Acc Pitch')
plt.title('Pitch vs Time')
plt.legend()
plt.grid(True)
plt.subplot(3,1,2)
plt.plot(time_data, gyro_roll_data, 'blue', label='Gyro Roll')
plt.plot(time_data, acc_roll_data,  'b--', label='Acc Roll')
plt.title('Roll vs Time')
plt.legend()
plt.grid(True)
plt.subplot(3,1,3)
plt.plot(time_data, gyro_roll_data, 'blue', label='Gyro Roll')
plt.title('Yaw vs Time')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Global lists to store the time stamps and sensor values.
time_data = []
gyro_pitch_data = []
gyro_roll_data  = []
gyro_yaw_data   = []
acc_pitch_data  = []
acc_roll_data   = []
comp_pitch_data = []
comp_roll_data  = []

# We'll use this to record the start time (in seconds)
start_time = None

def notification_handler(sender: int, data: bytearray):
    global start_time
    try:
        # Convert the received bytearray to a string and remove extra whitespace.
        message = data.decode("utf-8").strip()
        # Expected message format:
        # "Gyro Pitch: <gyro_pitch>; Gyro Roll: <gyro_roll>; Acc Pitch: <pitch_acc>; Acc Roll: <roll_acc>; CompPitch: <comp_pitch>; CompRoll: <comp_roll>"
        
        # Split the message into parts using the semicolon as the delimiter.
        tokens = message.split(';')
        if len(tokens) < 6:
            print("Incomplete data received:", message)
            return
        
        # Parse each token by splitting on the colon.
        gyro_pitch = float(tokens[0].split(':')[1].strip())
        gyro_roll  = float(tokens[1].split(':')[1].strip())
        gyro_yaw   = float(tokens[2].split(':')[1].strip())
        acc_pitch  = float(tokens[3].split(':')[1].strip())
        acc_roll   = float(tokens[4].split(':')[1].strip())
        comp_pitch = float(tokens[5].split(':')[1].strip())
        comp_roll  = float(tokens[6].split(':')[1].strip())
        
        # Record the time (in seconds since the first notification)
        if start_time is None:
            start_time = time.time()
        t = time.time() - start_time
        
        # Append the values to the global lists.
        time_data.append(t)
        gyro_pitch_data.append(gyro_pitch)
        gyro_roll_data.append(gyro_roll)
        gyro_yaw_data.append(gyro_yaw)
        acc_pitch_data.append(acc_pitch)
        acc_roll_data.append(acc_roll)
        comp_pitch_data.append(comp_pitch)
        comp_roll_data.append(comp_roll)
        
    except Exception as e:
        print("Error parsing notification:", e)
        print("Raw message:", message)
```

```python
# Connect to the Artemis Device
ble.connect()
ble.start_notify(ble.uuid['RX_STRING'], notification_handler)
for i in range(150):
    ble.send_command(CMD.GET_FROM_GYROSCOPE, "")
```

```python
# ble.stop_notify(ble.uuid['RX_STRING'])
plt.figure(figsize=(10, 8))

# Plot for Pitch
plt.subplot(3, 1, 1)
plt.plot(time_data, acc_pitch_data,  'g--', label='Acc Pitch (Filtered)')
plt.plot(time_data, gyro_pitch_data, 'r--', label='Gyro Pitch')
plt.plot(time_data, comp_pitch_data, 'b-',  label='Comp Pitch')
plt.xlabel('Time (s)')
plt.ylabel('Pitch (°)')
plt.title('Pitch vs. Time')
plt.legend()
plt.grid(True)

# Plot for Roll
plt.subplot(3, 1, 2)
plt.plot(time_data, acc_roll_data,  'g--', label='Acc Roll (Filtered)')
plt.plot(time_data, gyro_roll_data, 'r--', label='Gyro Roll')
plt.plot(time_data, comp_roll_data, 'b-',  label='Comp Roll')
plt.xlabel('Time (s)')
plt.ylabel('Roll (°)')
plt.title('Roll vs. Time')
plt.legend()
plt.grid(True)

plt.subplot(3,1,3)
plt.plot(time_data, gyro_yaw_data, 'r--', label='Gyro Yaw')
plt.title('Yaw vs Time')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
time_data = []
ax_data = []
ay_data = []
az_data = []
gx_data = []
gy_data = []
gz_data = []

# We'll parse lines of the form:
# "T:<time>|AX:<ax>|AY:<ay>|AZ:<az>|GX:<gx>|GY:<gy>|GZ:<gz>"
def imu_notification_handler(sender, data):
    message = data.decode('utf-8').strip()
    # Example: "T:1234|AX:0.01|AY:-0.02|AZ:9.81|GX:0.02|GY:-0.01|GZ:0.00"
    if not message.startswith("T:"):
        # It's possible we got a different message or partial data
        print("Unexpected data:", message)
        return
    try:
        # Split by '|'
        tokens = message.split('|')
        t_str   = tokens[0].split(':')[1]  # "1234"
        ax_str  = tokens[1].split(':')[1]  # "0.01"
        ay_str  = tokens[2].split(':')[1]  # "-0.02"
        az_str  = tokens[3].split(':')[1]  # "9.81"
        gx_str  = tokens[4].split(':')[1]  # "0.02"
        gy_str  = tokens[5].split(':')[1]  # "-0.01"
        gz_str  = tokens[6].split(':')[1]  # "0.00"
        # Convert to numeric
        t_val  = float(t_str)   # in ms from the Artemis side
        ax_val = float(ax_str)
        ay_val = float(ay_str)
        az_val = float(az_str)
        gx_val = float(gx_str)
        gy_val = float(gy_str)
        gz_val = float(gz_str)
        # Append to global arrays
        time_data.append(t_val)
        ax_data.append(ax_val)
        ay_data.append(ay_val)
        az_data.append(az_val)
        gx_data.append(gx_val)
        gy_data.append(gy_val)
        gz_data.append(gz_val)
    except Exception as e:
        print("Error parsing IMU data:", e)
        print("Raw message:", message)
```

```python
ble.start_notify(ble.uuid["RX_STRING"], imu_notification_handler)
# Trigger IMU Data Storage on the Arduino
print("Requesting Arduino to store IMU data for 5 seconds...")
ble.send_command(CMD.STORE_IMU_DATA, "")
```

```python
# Request the Data to be Sent
print("Requesting Arduino to send stored IMU data...")
ble.send_command(CMD.SEND_IMU_DATA, "")
```

```python
# NOTE: The time_data is in ms (from the Artemis). We can keep it in ms or convert to seconds.
time_s = [t/1000.0 for t in time_data]  # Convert ms -> s

plt.figure(figsize=(10, 8))

# Plot Accel
plt.subplot(2,1,1)
plt.plot(time_s, ax_data, 'r-', label='AX')
plt.plot(time_s, ay_data, 'g-', label='AY')
plt.plot(time_s, az_data, 'b-', label='AZ')
plt.xlabel('Time (s)')
plt.ylabel('Acceleration (g)')
plt.title('Accelerometer Data')
plt.grid(True)
plt.legend()

# Plot Gyro
plt.subplot(2,1,2)
plt.plot(time_s, gx_data, 'r-', label='GX')
plt.plot(time_s, gy_data, 'g-', label='GY')
plt.plot(time_s, gz_data, 'b-', label='GZ')
plt.xlabel('Time (s)')
plt.ylabel('Gyro (dps)')
plt.title('Gyroscope Data')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

# 8) Print some quick info
print(f"Received {len(time_data)} IMU samples from the Arduino.")
```
